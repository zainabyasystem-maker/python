```python
# 1. Print name and city
name = "Imama Zainab"
city = "Multan"
print(f"Name: {name}, City: {city}")

# 2. Variables with f-string
age = 20
print(f"My name is {name}, I am {age} years old and I live in {city}.")

# 3. Marks List
marks = [78, 85, 90, 72, 95]
print("Total:", sum(marks))
print("Average:", sum(marks) / len(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))

# 4. Average function
def average(numbers):
    return sum(numbers) / len(numbers)

print("Average via function:", average(marks))

# 5. Loop for Pass/Fail
for mark in marks:
    if mark >= 50:
        print(f"{mark}: Pass")
    else:
        print(f"{mark}: Fail")
```

```python
# 6. Product sales dictionary
sales_data = {"Product A": 45, "Product B": 60, "Product C": 55, "Product D": 30}
for product, units in sales_data.items():
    print(f"{product}: {units} units")
print("Total units sold:", sum(sales_data.values()))

# 7. Sales above 50 count
sales_list = [45, 60, 55, 30, 70, 80]
above_50 = sum(1 for sale in sales_list if sale > 50)
print(f"Months with sales above 50: {above_50}")

# 8. User marks input & grade
user_marks = float(input("Enter your marks: "))
if user_marks >= 90:
    grade = 'A'
elif user_marks >= 80:
    grade = 'B'
elif user_marks >= 70:
    grade = 'C'
else:
    grade = 'F'
print(f"Grade: {grade}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Monthly sales dataframe
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [45, 60, 55, 30, 70, 80]
})

print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())

# Bar Chart
df.plot(x="Month", y="Sales", kind="bar", title="Monthly Sales Overview")
plt.ylabel("Sales Units")
plt.show()
```
